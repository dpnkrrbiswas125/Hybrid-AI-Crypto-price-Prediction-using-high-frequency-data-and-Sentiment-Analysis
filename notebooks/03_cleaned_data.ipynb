```python
import pandas as pd 
import numpy as np  
df3 = pd.read_csv('/home/dipankar/Desktop/crypto_price_prediction/data/raw/BTCUSDT_5m_UTC_18-08-2011_to_21-08-2026.csv')
print(df3.head())
df3.shape
df3.info()
df3.describe()
```

```python
#Checking for the null and the duplicate values
print("Number of null values and duplicate values in the dataset:")
print(df3.isnull().sum())
print("Number of duplicate values in the dataset:")
print(df3.duplicated().sum())
```

```python
#If there is any null and the duplicate values just drop them from the dataset
df3.dropna()
df3.drop_duplicates()
```

```python
#Removing the unnecessary attributes from the dataset
df3.drop(['exchange','symbol','quote_currency','source_dataset','is_synthetic','quality_flag'], axis=1, inplace=True)
```

```python
#checking the columns of the dataset
print("Columns of the dataset:")
print(df3.columns)
```

```python
#Adding the log returns of the close price to the dataset
df3["log_return_5m"] = np.log(df3["close"] / df3["close"].shift(1))

df3["log_return_15m"] = np.log(df3["close"] / df3["close"].shift(3))

print("First 5 rows of the dataset with new features:")
print(df3.head(100))
```

```python
#Adding the Technical indicators to the dataset
#Adding the momentum, trend and volatility 
from ta.momentum import RSIIndicator
from ta.trend import MACD
from ta.volatility import BollingerBands
# RSI
rsi = RSIIndicator(df3['close'], window=14)
df3['rsi'] = rsi.rsi()

# MACD
macd = MACD(df3['close'], window_slow=26, window_fast=12, window_sign=9, fillna=False)
df3['macd'] = macd.macd()
# Bollinger Bands
bollinger = BollingerBands(df3['close'], window=20, window_dev=2)
df3['bollinger_hband'] = bollinger.bollinger_hband()
df3['bollinger_lband'] = bollinger.bollinger_lband()
df3['bollinger_mavg'] = bollinger.bollinger_mavg()
print("First 5 rows of the dataset with new features:")
print(df3.head(45))
```

```python
#Adding the target variable to the dataset
df3["target_log_return_5m"] = np.log(
    df3["close"].shift(-1) / df3["close"]
)
df3["target_log_return_15m"] = np.log(
    df3["close"].shift(-3) / df3["close"]
)
print("First 45 rows of the dataset with new features:")
print(df3.head(45))
```

```python
#Phase 7: Dropping the missing values from the dataset
df3 = df3.dropna().reset_index(drop=True)
df3 = df3.drop_duplicates().reset_index(drop=True)
print("Dataset after dropping missing values:")
print(df3.isna().sum())
print("Number of duplicate values in the dataset:")
print(df3.duplicated().sum())
print(df3.head(45))
```
