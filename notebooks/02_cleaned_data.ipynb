```python
import pandas as pd 
import numpy as np  
df1 = pd.read_csv('/home/dipankar/Desktop/crypto_price_prediction/data/raw/btc_5m_data.csv')
print(df1.head())
df1.info()
df1.describe()
```

```python
#Number of the missing values and duplicates in the dataset
print(df1.isnull().sum())
print(df1.duplicated().sum())
```

```python
df1.head()
df1 = df1.dropna()
print("Number of The missing values in the dataset after dropping:")
print(df1.isnull().sum())
#Dropping the duplicates from the dataset
df1 = df1.drop_duplicates()
print("Number of duplicates in the dataset after dropping:")
print(df1.duplicated().sum())
print("First 5 rows of the cleaned dataset:")
print(df1.head())
```

```python
#Adding the log returns of the close price to the dataset
df1["log_return_5m"] = np.log(df1["close"] / df1["close"].shift(1))

df1["log_return_15m"] = np.log(df1["close"] / df1["close"].shift(3))

print("First 5 rows of the dataset with new features:")
print(df1.head())
```

```python
#Phase 5: Adding the Technical indicators to the dataset
#Adding the momentum, trend and volatility 
from ta.momentum import RSIIndicator
from ta.trend import MACD
from ta.volatility import BollingerBands
# RSI
rsi = RSIIndicator(df1['close'], window=14)
df1['rsi'] = rsi.rsi()

# MACD
macd = MACD(df1['close'], window_slow=26, window_fast=12, window_sign=9, fillna=False)
df1['macd'] = macd.macd()
# Bollinger Bands
bollinger = BollingerBands(df1['close'], window=20, window_dev=2)
df1['bollinger_hband'] = bollinger.bollinger_hband()
df1['bollinger_lband'] = bollinger.bollinger_lband()
df1['bollinger_mavg'] = bollinger.bollinger_mavg()
print("First 5 rows of the dataset with new features:")
print(df1.head(45))
```

```python
#Phase 6: Adding the target variable to the dataset
df1["target_log_return_5m"] = np.log(
    df1["close"].shift(-1) / df1["close"]
)
df1["target_log_return_15m"] = np.log(
    df1["close"].shift(-3) / df1["close"]
)
print("First 45 rows of the dataset with new features:")
print(df1.head(45))
```

```python
#Phase 7: Dropping the missing values from the dataset
df1 = df1.dropna().reset_index(drop=True)
print("Dataset after dropping missing values:")
print(df1.isna().sum())
print(df1.head(45))
```

```python
#Phase 8: Saving the cleaned dataset to a new CSV file
df1.to_csv('/home/dipankar/Desktop/crypto_price_prediction/data/processed/btc_5m_cleaned_data.csv', index=False)
```
