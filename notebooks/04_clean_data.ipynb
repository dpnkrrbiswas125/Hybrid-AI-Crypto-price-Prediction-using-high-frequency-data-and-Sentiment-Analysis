```python
import pandas as pd 
import numpy as np  
df4 = pd.read_csv('/home/dipankar/Desktop/crypto_price_prediction/data/raw/BTCUSD-5m-2026-10-07.csv')
print(df4.head())
df4.shape
df4.info()
df4.describe()
```

```python
# Assign Binance Kline column names
df4.columns = [
    "open_time",
    "open",
    "high",
    "low",
    "close",
    "volume",
    "close_time",
    "quote_asset_volume",
    "number_of_trades",
    "taker_buy_base_asset_volume",
    "taker_buy_quote_asset_volume",
    "ignore"
]
```

```python
print(df4.columns)
df4.head()
```

```python
# Convert open_time from microseconds to UTC datetime
df4["Datetime"] = pd.to_datetime(
    df4["open_time"], unit="us", utc=True
)
```

```python
#Keeping only the required columns
df4 = df4[["Datetime", "open", "high", "low", "close", "volume"]]
```

```python
#checking for missing and duplicate values in the dataset
print("Missing values in the dataset:")
print(df4.isna().sum())
print("Duplicate values in the dataset:")
print(df4.duplicated().sum())
```

```python
#removing the missing and duplicate values from the dataset
df4.dropna()
df4.drop_duplicates()
```

```python
#Adding the log returns of the close price to the dataset
df4["log_return_5m"] = np.log(df4["close"] / df4["close"].shift(1))

df4["log_return_15m"] = np.log(df4["close"] / df4["close"].shift(3))

print("First 5 rows of the dataset with new features:")
print(df4.head(100))
```

```python
#Adding the Technical indicators to the dataset
#Adding the momentum, trend and volatility 
from ta.momentum import RSIIndicator
from ta.trend import MACD
from ta.volatility import BollingerBands
# RSI
rsi = RSIIndicator(df4['close'], window=14)
df4['rsi'] = rsi.rsi()

# MACD
macd = MACD(df4['close'], window_slow=26, window_fast=12, window_sign=9, fillna=False)
df4['macd'] = macd.macd()
# Bollinger Bands
bollinger = BollingerBands(df4['close'], window=20, window_dev=2)
df4['bollinger_hband'] = bollinger.bollinger_hband()
df4['bollinger_lband'] = bollinger.bollinger_lband()
df4['bollinger_mavg'] = bollinger.bollinger_mavg()
print("First 5 rows of the dataset with new features:")
print(df4.head(45))
```

```python
#Adding the target variable to the dataset
df4["target_log_return_5m"] = np.log(
    df4["close"].shift(-1) / df4["close"]
)
df4["target_log_return_15m"] = np.log(
    df4["close"].shift(-3) / df4["close"]
)
print("First 45 rows of the dataset with new features:")
print(df4.head(45))
```

```python
#Dropping the missing values from the dataset
df4 = df4.dropna().reset_index(drop=True)
df4 = df4.drop_duplicates().reset_index(drop=True)
print("Dataset after dropping missing values:")
print(df4.isna().sum())
print("Number of duplicate values in the dataset:")
print(df4.duplicated().sum())
print(df4.head(45))
```
